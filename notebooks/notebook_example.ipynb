## 1. Charger la configuration

`charger_config()` lit `config/defaut.yaml`, vérifie toutes les valeurs, puis renvoie un objet `cfg`. Si une valeur est invalide, une erreur `ConfigInvalide` indique la clé en cause.

```python
from railinsight.config import charger_config

cfg = charger_config()
cfg
```

```python
p = cfg.score_fiabilite

PERIODE_DEBUT = p.periode_debut  # "AAAA-MM", ou None = depuis le début
PERIODE_FIN = p.periode_fin  # "AAAA-MM", ou None = jusqu'à la fin
MIN_TRAINS = p.min_trains
POIDS = p.poids

GRAINE = cfg.graine  # graine commune à tout le projet

print("poids :", POIDS)
print("min_trains :", MIN_TRAINS)
print("période :", PERIODE_DEBUT, "->", PERIODE_FIN)
```
